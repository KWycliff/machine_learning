```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 25, 50)

# Linear function
k, n = 0.5, 2

y = k * x + n

# Noise
np.random.seed(12345)
noise = np.random.randn(y.size)

y_noise = y + noise
```

```python
plt.figure(figsize=(14, 8))
plt.plot(x, y, label='y_true', color='blue')
plt.plot(x, y_noise, '*', label='y_noise', color='red')
plt.title("Data Visualization")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Compute means
x_mean = np.mean(x)
y_mean = np.mean(y_noise)

# Compute slope (beta_1)
beta_1 = np.sum((x - x_mean) * (y_noise - y_mean)) / np.sum((x - x_mean) ** 2)

# Compute intercept (beta_0)
beta_0 = y_mean - beta_1 * x_mean

print("Slope :", beta_1)
print("Intercept :", beta_0)
```

```python
# Prediction
y_pred = beta_0 + beta_1 * x

# Calculate SSE (Sum of Squared Errors)
SSE = np.sum((y_noise - y_pred) ** 2)
print(f"Sum of Squared Errors (SSE): {SSE}")
print(f"Sum of Squared Errors (SSE) per sample: {SSE / y_noise.shape[0]}")
```

```python
plt.figure(figsize=(14, 8))
plt.plot(x, y, label='y_true', color='blue')
plt.plot(x, y_pred, label='y_pred', color='green')
plt.plot(x, y_noise, '*', label='y_noise', color='red')
plt.title("Linear Regression")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
