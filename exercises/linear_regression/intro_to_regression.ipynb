## ✏️ Task 3 — Train a linear regression model

Now let's train a model to predict the price from the size.

A **linear regression** model learns the equation of a line:

$$\text{price} = w \times \text{size} + b$$

where $w$ is the **slope** and $b$ is the **intercept**.

Follow these 3 steps:

**Step 1** — Prepare the data: separate the input (`X`) and the target (`y`)
```python
X = df[["size_m2"]]       # input (must use double brackets)
y = df["price_euro"]       # target
```

**Step 2** — Create and train the model
```python
model = LinearRegression()
model.fit(X, y)
```

**Step 3** — Print what the model learned
```python
print("Slope (w):", model.coef_[0])
print("Intercept (b):", model.intercept_)
```

````python
# ✏️ YOUR CODE HERE — Step 1: Prepare the data
X = df[['size_m2']]  # Features (independent variable)
y = df['price_euro']  # Target (dependent variable)

# ✏️ YOUR CODE HERE — Step 2: Create and train the model
model = LinearRegression()
model.fit(X, y)


# ✏️ YOUR CODE HERE — Step 3: Print the slope and intercept
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
````

### 👀 See the result — Run this cell to plot the regression line

If your model above is trained correctly, run the cell below to see the line your model found overlaid on the data.

````python
# Plot the data and the regression line (just run this cell)
plt.scatter(df["size_m2"], df["price_euro"], label="Data")
plt.plot(df["size_m2"], model.predict(df[["size_m2"]]), color="red", linewidth=2, label="Regression line")
plt.xlabel("Size (m²)")
plt.ylabel("Price (€)")
plt.title("Linear Regression: Size → Price")
plt.legend()
plt.show()
````

# Part 2 — Adding a second feature

In Part 1 we used only one feature to predict price. In Part 2, we add a second feature.

The file `housing_2d_stories.csv` has the same homes but with two features:
- **`area`** — the house area
- **`stories`** — the number of floors
- **`price`** — the sale price

The regression equation now becomes:

$$\text{price} = w_1 \times \text{area} + w_2 \times \text{stories} + b$$

## ✏️ Task 4 — Do it all again with 2 features!

Repeat the same steps as before, but this time:
1. Load `housing_2d_stories.csv` into a new DataFrame called `df2`
2. Prepare `X2` with **both** columns `["area", "stories"]` and `y2` with `"price"`
3. Create a new model, train it, and print the coefficients

**Hint for step 2:**
```python
X2 = df2[["area", "stories"]]
y2 = df2["price"]
```

````python
# ✏️ YOUR CODE HERE — Load housing_2d_stories.csv into df2 and display the first rows
df2 = pd.read_csv('housing_2d_stories.csv')
df2.head()
````

````python
# ✏️ YOUR CODE HERE — Prepare X2 (both features) and y2 (target)
X2 = df2[['area', 'stories']]  # Features (independent variables)
y2 = df2['price']  # Target (dependent variable)

# ✏️ YOUR CODE HERE — Create a new model, train it
model2 = LinearRegression()
model2.fit(X2, y2)

# ✏️ YOUR CODE HERE — Print the coefficients and intercept
print("Coefficients:", model2.coef_)
print("Intercept:", model2.intercept_)
````

### 👀 See the result — Run this cell to compare both models

If your model above is trained correctly, run the cell below to see a comparison of the predictions.

````python
# Compare predictions from both models (just run this cell)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left plot: model trained with area only (on df2)
model2_area_only = LinearRegression()
model2_area_only.fit(df2[["area"]], y2)
pred_area_only = model2_area_only.predict(df2[["area"]])
axes[0].scatter(y2, pred_area_only, alpha=0.5)
axes[0].plot([y2.min(), y2.max()], [y2.min(), y2.max()], "r--", linewidth=2)
axes[0].set_xlabel("Actual Price")
axes[0].set_ylabel("Predicted Price")
axes[0].set_title("1 feature (area only)")

# Right plot: 2-feature model
axes[1].scatter(y2, model2.predict(X2), alpha=0.5, color="green")
axes[1].plot([y2.min(), y2.max()], [y2.min(), y2.max()], "r--", linewidth=2)
axes[1].set_xlabel("Actual Price")
axes[1].set_ylabel("Predicted Price")
axes[1].set_title("2 features (area + stories)")

plt.tight_layout()
plt.show()

print("💡 The closer the points are to the red dashed line, the better the predictions!")
````

# 📝 Homework — Which model is better?

On the same dataset (`housing_2d_stories.csv`), compare:
- Model A: uses only `area`
- Model B: uses `area` and `stories`

Your task: come up with a way to measure how far off each model's predictions are from the real prices.

Some guidelines:
- Use the same target (`y2`) for both models
- You can get predictions with `.predict(...)`
- Compute a single number that summarises error (for example MAE)
- Compare that number for both models — which one wins?

💡 *There is no single right answer — be creative and try things out!*

````python
# ✏️ YOUR CODE HERE — Compare area-only vs area+stories on the same dataset
# compare using the RMSE 
from sklearn.metrics import mean_squared_error
import numpy as np
mse_area_only = mean_squared_error(y2, pred_area_only)
mse_area_stories = mean_squared_error(y2, model2.predict(X2))
rmse_area_only = np.sqrt(mse_area_only)
rmse_area_stories = np.sqrt(mse_area_stories)
print(f"Root Mean Squared Error (area only): {rmse_area_only:.2f}")
print(f"Root Mean Squared Error (area + stories): {rmse_area_stories:.2f}")
````
