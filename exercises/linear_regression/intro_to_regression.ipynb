$$\text{price} = w \times \text{size} + b$$

where $w$ is the **slope** and $b$ is the **intercept**.


**Step 3** — Print what the model learned

### 👀 See the result — Run this cell to compare both models

````python
# Compare predictions from both models (just run this cell)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left plot: model trained with area only (on df2)
model2_area_only = LinearRegression()
model2_area_only.fit(df2[["area"]], y2)
pred_area_only = model2_area_only.predict(df2[["area"]])
axes[0].scatter(y2, pred_area_only, alpha=0.5)
axes[0].plot([y2.min(), y2.max()], [y2.min(), y2.max()], "r--", linewidth=2)
axes[0].set_xlabel("Actual Price")
axes[0].set_ylabel("Predicted Price")
axes[0].set_title("1 feature (area only)")

# Right plot: 2-feature model
axes[1].scatter(y2, model2.predict(X2), alpha=0.5, color="green")
axes[1].plot([y2.min(), y2.max()], [y2.min(), y2.max()], "r--", linewidth=2)
axes[1].set_xlabel("Actual Price")
axes[1].set_ylabel("Predicted Price")
axes[1].set_title("2 features (area + stories)")

plt.tight_layout()
plt.show()

print("💡 The closer the points are to the red dashed line, the better the predictions!")
````
